```python
# Распределение значений категориальных признаков в train и test

for i, col in enumerate(cat_cols):
    fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(12, 6))
    fig.suptitle(f'Распределение {col}', fontsize=10, ha='center')

    for j, df in enumerate([train, test]):
        vals = df[col].value_counts()
        colors = sns.color_palette('muted', n_colors=len(vals))
        
        # Pie
        wedges, texts, autotexts = ax[j, 0].pie(x=vals,
                                    labels=vals.index,
                                    colors=colors,
                                    autopct=lambda n, vals=vals: f'{n:.2f}% ({n * vals.sum() / 100:.0f})',
                                    textprops={'fontsize': 7},
                                    # labeldistance=1.2,
                                    pctdistance=0.6,
                                    rotatelabels=False)
        
        for text in texts:
            text.set_fontsize(7)
            # text.set_horizontalalignment('center')
        for autotext in autotexts:
            autotext.set_fontsize(7)
            autotext.set_horizontalalignment('center')
            autotext.set_verticalalignment('center')
    
        ax[j, 0].set_title(f"{'Train' if j==0 else 'Test'} - {col}", fontsize=8)
        
        # Bar
        ax[j, 1].bar(x=vals.index, 
                     height=vals.values, 
                     color=colors)
        
        ax[j, 1].set_title(f"{'Train' if j==0 else 'Test'} - {col}", fontsize=8)
        ax[j, 1].tick_params(axis='x', rotation=0, labelsize=7)
        ax[j, 1].tick_params(axis='y', labelsize=7)


    plt.tight_layout()

    plt.show();
```

## Предобработка признаков

```python
# Преобразование признаков

def prep_features(df, bins=None):
    bin_cols = ['Sex', 'HasPartner', 'HasChild', 'IsBillingPaperless']
    
    tmp = df.copy()

    # Бинаризируем признаки
    for col in bin_cols:
        if col == 'Sex':
            tmp[col] = (tmp[col] == 'Male').astype(int)
        else:
            tmp[col] = (tmp[col] == 'Yes').astype(int)

    # Сгруппируем с разделением на 10 групп периодов
    # tmp['ClientPeriodGroup'] = pd.cut(tmp['ClientPeriod'], bins=10, labels=False) # Правильно применить к test нужно

    # Сгруппируем с разделением на 10 групп периодов

    # Для трейн
    if bins is None:
        tmp['ClientPeriodGroup'], bins_edges = pd.cut(tmp['ClientPeriod'], 
                                                       bins=10, 
                                                       labels=False, 
                                                       retbins=True)
        return tmp, bins_edges

    # Для теста разбиение по интрвалам разбиения трейна 
    else:
        # bins = np.array([-0.072,7.2, 14.4, 21.6, 28.8, 36. , 43.2, 50.4, 57.6, 64.8, 72. ])
        
        tmp['ClientPeriodGroup'] = pd.cut(tmp['ClientPeriod'], 
                                          bins=bins, 
                                          labels=False, 
                                          include_lowest=True)
        return tmp
```

```python
# Создание новых признаков

def create_features(df):
    cols = ['HasPhoneService', 'HasInternetService','HasOnlineSecurityService', 
            'HasOnlineBackup', 'HasDeviceProtection', 'HasTechSupportAccess', 
            'HasOnlineTV', 'HasMovieSubscription']
    
    tmp = df.copy()

    not_test = False

    if 'Churn' in tmp.columns:
        not_test = True
        churn = tmp['Churn'].copy()
        tmp = tmp.drop('Churn', axis=1)

    # Средние траты за месяц
    tmp['AverageMonthlySpend'] = tmp['TotalSpent'] / (tmp['ClientPeriod'] + 0.0001)
    
    # Наличие семьи (сервисы на несколько человек)
    tmp['NotAlone'] = ((tmp['HasPartner'] == 1) | (tmp['HasChild'] == 1)).astype(int)

    # Одинокие пенсионеры
    tmp['IsAloneSenior'] = ((tmp['IsSeniorCitizen'] == 1) & (tmp['HasPartner'] == 0)).astype(int)

    # Количество сервисов
    mask = pd.DataFrame()
    for col in cols:
        if col == 'HasInternetService':
            mask[col] = (tmp[col] != 'No')
        else:
            mask[col] = (tmp[col] == 'Yes')
    tmp['HasMultipleServices'] = mask.sum(axis=1)

    # Возвращаем целевую переменную
    if not_test:
        tmp['Churn'] = churn
    
    return tmp
```

```python
def objective_rf(trial):
    params = {
        'n_estimators':         trial.suggest_int('n_estimators', 100, 500, step=50),
        'max_depth':            trial.suggest_int('max_depth', 3, 15),
        'min_samples_split':    trial.suggest_int('min_samples_split', 2, 20),
        'min_samples_leaf':     trial.suggest_int('min_samples_leaf', 1, 10),
        'max_features':         trial.suggest_categorical('max_features', ['sqrt', 'log2', None]),
        'class_weight':         trial.suggest_categorical('class_weight', ['balanced', 'balanced_subsample', None]),
        'random_state':         RANDOM_STATE,
        'n_jobs':               -1
        }
    
    X_train_rf, X_val_rf, y_train_rf, y_val_rf = train_test_split(X_train_stack,
                                                                  y_train_stack,
                                                                  test_size=0.25,
                                                                  random_state=RANDOM_STATE,
                                                                  stratify=y_train_stack
                                                                  )
    
    preprocessor = ColumnTransformer([
        ('num', StandardScaler(), num_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
    ])
    
    X_train_processed = preprocessor.fit_transform(X_train_rf)
    X_val_processed = preprocessor.transform(X_val_rf)
        
    model = RandomForestClassifier(**params)
    model.fit(X_train_processed, y_train_rf)
    
    y_pred = model.predict_proba(X_val_processed)[:, 1]
    
    return roc_auc_score(y_val_rf, y_pred)
```
