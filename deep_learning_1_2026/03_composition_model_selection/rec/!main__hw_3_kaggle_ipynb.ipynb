```python
def preprocess_df(df, binary_cols, multi_cat_cols, num_cols):
    """
    Универсальная предобработка:
    - бинарные колонки → 0/1 (Yes/No → 1/0, 'No internet service'/'No phone service' → 0)
    - категориальные колонки → one-hot
    - проверка наличия числовых колонок
    """
    df_proc = df.copy()

    # --- бинарные колонки ---
    for col in binary_cols:
        if col in df_proc.columns:
            # Заменяем текстовые значения на 0/1
            df_proc[col] = df_proc[col].replace({
                'Yes': 1,
                'No': 0,
                'No internet service': 0,
                'No phone service': 0,
                'Male': 1,
                'Female': 0
            })
            # Принудительно в int
            df_proc[col] = df_proc[col].astype(int)

    # --- мультикатегориальные колонки ---
    cat_to_encode = [c for c in multi_cat_cols if c in df_proc.columns]
    if cat_to_encode:
        df_proc = pd.get_dummies(df_proc, columns=cat_to_encode, drop_first=True)
        df_proc= df_proc.replace({True: 1, False: 0})

    # --- числовые колонки ---
    num_cols_exist = [c for c in num_cols if c in df_proc.columns]

    return df_proc, num_cols_exist
```

```python
# Списки признаков
binary_cols = [
    'HasPartner', 'HasChild', 'HasPhoneService', 'IsBillingPaperless',
    'HasOnlineSecurityService', 'HasOnlineBackup', 'HasDeviceProtection',
    'HasTechSupportAccess', 'HasOnlineTV', 'HasMovieSubscription',
    'Sex', 'HasMultiplePhoneNumbers'
]

multi_cat_cols = ['HasInternetService', 'HasContractPhone', 'PaymentMethod']

num_cols = ['ClientPeriod', 'MonthlySpending', 'TotalSpent']

# Применяем preprocess_df
df_tr_proc, num_cols_exist = preprocess_df(df_train, binary_cols, multi_cat_cols, num_cols)

# Проверяем результат
df_tr_proc.head()
```

```python
#Работаем с числовыми признаками, анализируем их распределение
def plot_numeric_distributions(df, num_cols):
    """
    Функция строит гистограммы с KDE и boxplots для числовых колонок,
    а также выводит базовую статистику.

    Параметры:
    -----------
    df : pandas.DataFrame
        DataFrame с данными.
    num_cols : list
        Список числовых колонок.
    """

    # Проверяем, какие колонки реально есть
    num_cols_exist = [col for col in num_cols if col in df.columns]
    if len(num_cols_exist) == 0:
        print("Нет числовых колонок для отображения.")
        return

    # ===== 0. Базовая статистика =====
    print("=== Базовая статистика ===")
    display(df[num_cols_exist].describe())

    # ===== 1. Гистограммы с KDE =====
    n_cols = 2  # графиков в строке
    n_rows = int(np.ceil(len(num_cols_exist)/n_cols))

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(12, 4*n_rows))
    axes = axes.flatten()

    for i, col in enumerate(num_cols_exist):
        sns.histplot(df[col], kde=True, bins=30, ax=axes[i])
        axes[i].set_title(f'Гистограмма и KDE: {col}')
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Count')

    # Убираем пустые подграфики
    for j in range(i+1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()

    # ===== 2. Boxplots =====
    fig, axes = plt.subplots(1, len(num_cols_exist), figsize=(5*len(num_cols_exist), 4))

    # если только одна колонка
    if len(num_cols_exist) == 1:
        axes = [axes]

    for i, col in enumerate(num_cols_exist):
        sns.boxplot(y=df[col], ax=axes[i])
        axes[i].set_title(f'Boxplot: {col}')
        axes[i].set_ylabel(col)

    plt.tight_layout()
    plt.show()
```

```python
#Работаем с числовыми признаками, анализируем их распределение
from scipy.stats import skew
from sklearn.preprocessing import QuantileTransformer

def compute_skewness(df, cols):
    """
    Вычисляет skewness для списка числовых колонок и возвращает словарь.
    """
    skew_dict = {}
    for col in cols:
        if col in df.columns:
            skew_val = skew(df[col].dropna())
            skew_dict[col] = skew_val
            print(f"{col}: skewness = {skew_val:.2f}")
    return skew_dict

def log_transform(df, col, new_col=None):
    """
    Применяет логарифмирование log(1 + x) к колонке.
    """
    if new_col is None:
        new_col = f"{col}_log"
    if col in df.columns:
        df[new_col] = np.log1p(df[col])
        plt.figure(figsize=(6,3))
        sns.histplot(df[new_col], kde=True, bins=30)
        plt.title(f'Распределение log(1 + {col})')
        plt.show()
        print(f"Skewness после логарифма {new_col}: {skew(df[new_col].dropna()):.2f}")
    return df

def quantile_transform(df, col, new_col=None):
    """
    Применяет квантильное преобразование (QuantileTransformer) к колонке.
    """
    if new_col is None:
        new_col = f"{col}_quantile"
    if col in df.columns:
        qt = QuantileTransformer(output_distribution='normal')
        df[new_col] = qt.fit_transform(df[[col]])
        plt.figure(figsize=(6,3))
        sns.histplot(df[new_col], kde=True, bins=30)
        plt.title(f'Распределение {new_col} после Quantile Transformer')
        plt.show()
        print(f"Skewness после Quantile Transformer {new_col}: {skew(df[new_col].dropna()):.2f}")
    return df
```

```python
# 1. Считаем skew для всех числовых колонок
skews = compute_skewness(df_tr_proc, num_cols)
```

```python
#ClientPeriod и MonthlySpending → распределение почти симметричное, логарифмировать не нужно
#TotalSpent → есть положительная скошенность (skew ≈ 0.96), можно рассмотреть log, чтобы сгладить хвост

# 2. Логарифмирование TotalSpent
df_tr_proc = log_transform(df_tr_proc, 'TotalSpent')
```

```python
#TotalSpent_log → skew ≈ -0.74, сгладил длинный правый хвост
# Применим квантильное преобразование (Quantile Transformer), gреобразует распределение в нормальное, то есть разглаживает хвосты и выравнивает распределение
df_tr_proc = quantile_transform(df_tr_proc, 'TotalSpent')
```

```python
# Предобработка данных
df_tr_proc, num_cols_exist = preprocess_df(df_tr, binary_cols, multi_cat_cols, num_cols)

# Разделяем на признаки и целевой признак
X = df_tr_proc.drop(columns=['Churn'])
y = df_tr_proc['Churn']

# Импорт необходимых модулей
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

# Preprocessing pipeline
# Масштабируем только существующие числовые признаки
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols_exist)
    ],
    remainder='passthrough'  # все бинарные и one-hot признаки оставляем без изменений
)

# Полный pipeline
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000, solver='liblinear', random_state=42))
])

# ===== Сетка гиперпараметров =====
param_grid = {'classifier__C': [0.001, 0.01, 0.1, 1, 10, 100]}

# ===== GridSearchCV для выбора лучшего C по ROC-AUC =====
grid = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)

# ===== Обучение модели =====
grid.fit(X, y)

# ===== Результаты =====
print("Лучший C:", grid.best_params_['classifier__C'])
print("Лучший ROC-AUC (CV):", grid.best_score_)


# Сохраняем лучшую модель (вместе с pipeline)
joblib.dump(grid.best_estimator_, "best_logreg_pipeline.pkl")

print("Лучшая модель сохранена в 'best_logreg_pipeline.pkl'")
```
