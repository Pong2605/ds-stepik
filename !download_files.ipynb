# Скачивание по ссылке url

```python
import os
import requests

URL = 'https://zenodo.org/record/1199620/files/SO_vectors_200.bin?download=1'
FOLDER = 'content'
PATH = os.path.join(FOLDER, 'SO_vectors_200.bin')

# Проверка наличия папки и создание по необходимости
os.makedirs(FOLDER, exist_ok=True)

# Проверка наличия файла и скачивание по необходимости
if not os.path.exists(PATH):
    r = requests.get(URL)
    with open(PATH, 'wb') as f:
        f.write(r.content)
    print('Файл скачан:', PATH)
else:
    print('Файл уже скачан')
```

# Скачивание по ID файла в Google Drive

```python
import gdown
import zipfile

# Скачивание

FILE_ID = '1QqT4D0EoqJTy7v9VrNCYD-m964XZFR7_'
FOLDER = 'content'
PATH = os.path.join(FOLDER, 'stackoverflow_similar_questions.zip')

# Проверка наличия файла и скачивание по необходимости
if not os.path.exists(PATH):
    gdown.download(f'https://drive.google.com/uc?id={FILE_ID}', PATH, quiet=True)
    print('Файл скачан:', PATH)
else:
    print('Файл уже скачан')
```

# Распаковка

```python
# Распаковка

EXTRACT_FOLDER = 'content/stackoverflow_similar_questions'

if not os.path.exists(EXTRACT_FOLDER):
    with zipfile.ZipFile(PATH, 'r') as zf:
        zf.extractall(path=EXTRACT_FOLDER)
    print('Файл распакован в:', EXTRACT_FOLDER)
else:
    print('Файл уже распакован')
```

```python
import tarfile

EXTRACT_DIR = os.path.join(FOLDER, "aclImdb")

if os.path.exists(EXTRACT_DIR):
    print("Уже распаковано:", EXTRACT_DIR)
else:
    print("Распаковываю...")
    with tarfile.open(PATH, "r:gz") as tar:
        tar.extractall(path=FOLDER)
    print("Готово:", EXTRACT_DIR)
```
