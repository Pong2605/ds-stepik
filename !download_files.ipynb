# Скачивание по ссылке url

```python
import os
import requests

URL = 'http://ai.stanford.edu/~amaas/data/sentiment/aclImdb_v1.tar.gz'
FOLDER = 'content/IMDB'
FILE = os.path.join(FOLDER, 'aclImdb_v1.tar.gz')

# Создаём папку, если её нет
os.makedirs(FOLDER, exist_ok=True)

# Проверяем, есть ли файл
if os.path.exists(FILE):
    print('Файл уже есть:', FILE)
else:
    print('Скачиваю...')
    r = requests.get(URL)
    with open(FILE, 'wb') as f:
        f.write(r.content)
    print('Готово:', FILE)
```

# Распаковка

```python
import tarfile

EXTRACT_DIR = os.path.join(FOLDER, "aclImdb")

if os.path.exists(EXTRACT_DIR):
    print("Уже распаковано:", EXTRACT_DIR)
else:
    print("Распаковываю...")
    with tarfile.open(FILE, "r:gz") as tar:
        tar.extractall(path=FOLDER)
    print("Готово:", EXTRACT_DIR)
```
