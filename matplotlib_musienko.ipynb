```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
y_upper = y + 0.3
y_lower = y - 0.3

plt.figure(figsize=(12, 6))

# Основная линия
plt.plot(x, y, 'b-', linewidth=2, label='sin(x)')

# Заливка между верхней и нижней границами
plt.fill_between(x, y_lower, y_upper,
                 color='blue',
                 alpha=0.2,
                 label='Доверительный интервал')

plt.title('Заливка между кривыми')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x = np.linspace(0, 10, 100)

y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(12, 6))

plt.plot(x, y1, 'b-', linewidth=2, label='Кривая 1: sin(x)')
plt.plot(x, y2, 'r-', linewidth=2, label='Кривая 2: cos(x)')

# Заливка между кривыми с условием
plt.fill_between(x, y1, y2,
                 where=y1 > y2,  # Условие закраски
                 color='green',
                 alpha=0.4,
                 interpolate=True,
                 label='Кривая 1 выше Кривой 2')

plt.title('Заливка между кривыми')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
np.random.seed(42)
salaries = np.random.normal(50000, 15000, 1000).astype(int)
density = stats.gaussian_kde(salaries)
x_range = np.linspace(salaries.min(), salaries.max(), 200)

# График 1. Гистограмма
plt.hist(salaries, bins=30, density=True, alpha=0.5, color='skyblue', edgecolor='black', label='Гистограмма (нормированная)')

# График 2. График плотности
plt.plot(x_range, density(x_range), color='darkred', lw=2.5, label='График плотности (KDE)')
plt.fill_between(x_range, density(x_range), color='red', alpha=0.1)

plt.title('Совместная визуализация (Гистограмма + KDE)')
plt.xlabel('Зарплата (руб.)')
plt.ylabel('Плотность')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 3.2 Ящик с усами

```python
np.random.seed(42)

# Создаем данные с нужными характеристиками
data = np.concatenate([
    np.random.normal(50, 4, 80),      # основная масса данных (80 точек)
    np.random.normal(35, 4, 3),       # группа низких значений (5 точек)
    np.random.normal(65, 4, 3)        # группа высоких значений (5 точек)
])

# Добавляем несколько явных выбросов

box = plt.boxplot(data,
                  vert=True,  # вертикальная ориентация
                  patch_artist=True,  # закрашиваем ящики
                  notch=True,  # включаем выемку у медианы

                  boxprops=dict(facecolor='lightblue',
                                edgecolor='navy',
                                linewidth=2,
                                linestyle='-'),

                  # Настройка медианы
                  medianprops=dict(color='red',
                                   linewidth=3,
                                   linestyle='-'),

                  # Настройка усов
                  whiskerprops=dict(color='darkgreen',
                                    linewidth=2,
                                    linestyle='--'),

                  # Настройка засечек на усах
                  capprops=dict(color='purple',
                                linewidth=2,
                                linestyle='-'),

                  # Настройка выбросов
                  flierprops=dict(marker='o',
                                  markerfacecolor='orange',
                                  markersize=8,
                                  alpha=0.7,
                                  markeredgecolor='darkred',
                                  markeredgewidth=1.5))
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
sepal_length = iris.data[:, 0]

# Построение ящика с усами
plt.figure(figsize=(8, 6))
plt.boxplot(sepal_length, vert=True, patch_artist=True,
            boxprops=dict(facecolor='lightblue'),
            medianprops=dict(color='red', linewidth=2),
            whiskerprops=dict(color='blue', linestyle='--'),
            capprops=dict(color='darkblue'),
            flierprops=dict(marker='o', markerfacecolor='red', markersize=8, alpha=0.5))

plt.title('Ящик с усами для длины чашелистика (Iris)')
plt.ylabel('Длина чашелистика (см)')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
iris = load_iris()
setosa = iris.data[iris.target == 0, 0]
versicolor = iris.data[iris.target == 1, 0]
virginica = iris.data[iris.target == 2, 0]


# Сохраняем результат в переменную
box = plt.boxplot([setosa, versicolor, virginica],  # Передаем сразу все данные в одном списке
            patch_artist=True,
            tick_labels=['Setosa', 'Versicolor', 'Virginica'],
            notch=True, showmeans=True)

# Теперь box содержит все элементы графика
# box['boxes'] — список ящиков
# box['medians'] — список линий медиан
# box['whiskers'] — список усов и т.д.

# Перебираем ящики и настраиваем каждый отдельно
box['boxes'][0].set_facecolor('lightgreen')
box['boxes'][1].set_facecolor('lightskyblue')
box['boxes'][2].set_facecolor('lightcoral')

plt.title('Сравнение длины чашелистика для трех видов ирисов')
plt.ylabel('Длина чашелистика (см)')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Создаём данные для графиков
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

data = np.random.randn(1000)

categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]

# Создаём сетку 2×2 (4 графика)
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))

# Теперь в переменной fig — вся фигура
# axes — это двумерный массив (2 строки, 2 столбца) с объектами осей

print(fig)  # Figure(640x480)
# Одно полотно для 4 графиков

print(axes)
# Вывод:
# [[<Axes: > <Axes: >]
#  [<Axes: > <Axes: >]]
# 4 слота для графиков

# Строим график в левом верхнем углу
axes[0, 0].plot(x, y1, color='red', linewidth=2)
axes[0, 0].set_title('График 1: Синус')
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('sin(x)')
axes[0, 0].grid(True, alpha=0.3)

# Правый верхний
axes[0, 1].scatter(x[::5], y2[::5], color='blue', alpha=0.6, s=30)  # берём каждый 5-й элемент для разреженности
axes[0, 1].set_title('График 2: Косинус (точечный)')
axes[0, 1].set_xlabel('x')
axes[0, 1].set_ylabel('cos(x)')
axes[0, 1].grid(True, alpha=0.3)

# Левый нижний
axes[1, 0].hist(data, bins=30, color='green', alpha=0.7, edgecolor='black')
axes[1, 0].set_title('График 3: Гистограмма')
axes[1, 0].set_xlabel('Значение')
axes[1, 0].set_ylabel('Частота')
axes[1, 0].grid(True, alpha=0.3)

# Правый нижний
axes[1, 1].bar(categories, values, color='purple', alpha=0.7, edgecolor='black')
axes[1, 1].set_title('График 4: Столбчатая диаграмма')
axes[1, 1].set_xlabel('Категория')
axes[1, 1].set_ylabel('Значение')
axes[1, 1].grid(axis='y', alpha=0.3)

fig.suptitle('Анализ данных математических функций', fontsize=16, fontweight='bold')

plt.tight_layout()

plt.show()
```
