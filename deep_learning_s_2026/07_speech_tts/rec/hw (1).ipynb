```python
from transformers import Wav2Vec2FeatureExtractor, WavLMForXVector
from datasets import load_dataset
import torch
import torchaudio

feature_extractor = Wav2Vec2FeatureExtractor.from_pretrained('microsoft/wavlm-base-sv')
model = WavLMForXVector.from_pretrained('microsoft/wavlm-base-sv')
```

```python
# 3 различных варианта и подсчитайте similarity этих записей друг с другом
```

```python
def load_audio(audio_path):
    a, sr = torchaudio.load(audio_path)
    if sr != 16000:
        resample_rate = 16000
        resampler = torchaudio.transforms.Resample(sr, resample_rate, dtype=a.dtype)
        resampled_waveform = resampler(a)
        a = resampled_waveform
    return a

def extract_embedding(audio_path):
    audio = load_audio(audio_path)
    inputs = feature_extractor(audio[0], sampling_rate=16000, return_tensors="pt")
    embeddings = model(**inputs).embeddings
    embeddings = torch.nn.functional.normalize(embeddings, dim=-1).cpu()
    return embeddings[0]

def cosine_similarity(emb1, emb2):
    cosine_sim = torch.nn.CosineSimilarity(dim=-1)
    similarity = cosine_sim(emb1, emb2)
    return similarity

def is_speakers_same(similarity, threshold = 0.86):
    print(round(similarity.item(), 4))
    if similarity.item() < threshold:
        print("Speakers are not the same!")
    else:
        print("Speakers are the same!")
```

```python
emb1 = extract_embedding(os.path.join(test_wavs_dir, 'test_samp_0.wav'))
emb2 = extract_embedding(os.path.join(test_wavs_dir, 'test_samp_1.wav'))
emb3 = extract_embedding(os.path.join(test_wavs_dir, 'test_samp_2.wav'))
```

```python
sim_1_2 = cosine_similarity(emb1, emb2)
sim_1_3 = cosine_similarity(emb1, emb3)
sim_3_2 = cosine_similarity(emb3, emb2)

is_speakers_same(sim_1_2, threshold = 0.86)
is_speakers_same(sim_1_3, threshold = 0.86)
is_speakers_same(sim_3_2, threshold = 0.86)
```

```python
mean_val = (sim_1_2 + sim_1_3 + sim_3_2) / 3
```

```python
# подсчитайте similarity этих записей со всеми сгенерированными
```

```python
gt_path = test_wavs_dir
zero_shot_path = 'results/zero_shot'
ft_1min_path = 'results/finetuned_1min'
ft_full_path = 'results/finetuned_full'
cnt_audio = 10
```

```python
def get_mean_similarity(gen_audios_path, gt_path):
    sim_vals = []
    for i in range(cnt_audio):
        emb_orig = extract_embedding(os.path.join(gt_path, f'test_samp_{i}.wav'))
        emb_generated = extract_embedding(os.path.join(gen_audios_path, f'sent_{i}.wav'))
        sim = cosine_similarity(emb_orig, emb_generated)
        sim_vals.append(sim)
        is_speakers_same(sim, threshold = 0.86)

    mean_sim = torch.stack(sim_vals).mean()
    return mean_sim
```

```python
mean_sim_zero_shot = get_mean_similarity(zero_shot_path, gt_path)
```
