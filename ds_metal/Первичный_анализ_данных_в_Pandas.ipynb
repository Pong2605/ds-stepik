```python
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
```

## Генерируем банковский датасет

Создадим реалистичные данные о клиентах, как будто нам прислали выгрузку из банковской CRM.

```python
np.random.seed(42)
N = 50000
names = ['Иван', 'Ольга', 'Дмитрий', 'Екатерина', 'Сергей', 'Анна', 'Алексей', 'Мария', 'Николай', 'Татьяна']

df = pd.DataFrame({
    'client_id': range(1001, 1001 + N),
    'name': np.random.choice(names, N),
    'age': np.random.randint(21, 66, N),
    'income': np.random.randint(20000, 500001, N),
    'dti': np.round(np.random.uniform(0, 1, N), 2),
    'default': np.random.choice([True, False], N, p=[0.15, 0.85]),
    'application_date': pd.date_range('2026-01-01', periods=N, freq='h')
})

print(f"Сгенерировано строк: {len(df)}")
```

## head() и tail() — края таблицы

head показывает первые строки. tail — последние. По умолчанию — 5 строк, но можно указать любое число.

```python
print("Первые 5 строк:")
print(df.head())

print("\nПервые 10 строк:")
print(df.head(10))

print("\nПоследние 5 строк:")
print(df.tail())

print("\nВсе строки, кроме последних 3 (head(-3)):")
print(df.head(-3))

print("\nВсе строки, кроме первых 3 (tail(-3)):")
print(df.tail(-3))
```

## info() — полная сводка о структуре

Показывает количество непустых значений, типы данных и занимаемую память.


```python
print("Стандартный info():")
df.info()
```

## describe() — статистический разрез

По умолчанию — только числовые колонки (count, mean, std, min, 25%, 50%, 75%, max).

```python
print("describe() — только числовые колонки:")
print(df.describe())

print("\ndescribe(include='all') — все колонки (включая текст):")
print(df.describe(include='all'))

print("\ndescribe() с параметром percentiles для своих перцентилей:")
print(df.describe(percentiles=[0.1, 0.5, 0.9]))
```

## shape, columns, dtypes — быстрые атрибуты

Обрати внимание: это атрибуты, а не методы, поэтому скобки не нужны.

```python
print(f"shape (строки, колонки): {df.shape}")
print(f"Количество строк: {df.shape[0]}")
print(f"Количество колонок: {df.shape[1]}")
print(f"\nНазвания колонок:\n{df.columns.tolist()}")
print(f"\nТипы данных:\n{df.dtypes}")
```

## value_counts() — частоты значений

Самый быстрый способ узнать распределение категориальной переменной.

```python
print("Распределение по флагу default:")
print(df['default'].value_counts())

print("\nТо же в процентах (normalize=True):")
print(df['default'].value_counts(normalize=True) * 100)

print("\nРаспределение по именам:")
print(df['name'].value_counts())

print("\nТоп-3 самых частых имени:")
print(df['name'].value_counts().head(3))
```

##sample() — случайный срез

Полезно, когда данные отсортированы по дате и head/tail не дают полной картины.

```python
print("5 случайных строк:")
print(df.sample(5, random_state=42))

print("\n1% случайных строк:")
print(df.sample(frac=0.01, random_state=42))
```

##  Быстрый поиск проблем

Пропущенные значения, дубликаты и типы колонок — 3 вещи, которые нужно проверять сразу после загрузки.

```python
print("Пропущенные значения в каждой колонке:")
print(df.isnull().sum())

print("\nКоличество строк-дубликатов:")
print(df.duplicated().sum())

print("\nТолько числовые колонки:")
print(df.select_dtypes(include='number').columns.tolist())
```
