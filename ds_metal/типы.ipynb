2️⃣ **Числа с плавающей точкой (float)**

```python
# Переменная pi = число с точкой 3.1415926535. Это float.
pi = 3.1415926535
print("Число π примерно:", pi)

radius = 5.5  # тоже float
# Вычисляем площадь круга: π * R²
area = pi * radius ** 2  # сначала возведение в степень (**), потом умножение
# round(area, 2) — округляет число area до двух знаков после запятой.
print("Площадь круга радиусом", radius, "равна:", round(area, 2))

# / — обычное деление, всегда даёт float (даже если делится нацело).
print("10 / 3 =", 10 / 3)   # 3.333333...
# // — целочисленное деление (отбрасывает дробную часть). Результат — int.
print("10 // 3 =", 10 // 3) # 3
# % — остаток от деления (тоже int для целых чисел).
# Но это мы уже видели.

my_height = 1.75  # Метры. Меняйте число.
print("Мой рост:", my_height, "м")
# Умножаем рост в метрах на 39.37 (приблизительно дюймов в метре). Получаем float.
print("Рост в дюймах (примерно):", my_height * 39.37)
print()  # пустая строка
```

🔟 Байты (bytes)

```python
# b в начале означает, что это литерал байтов (не строка).
b_data = b"Hello, Robot"
print("Байты:", b_data)
# type() — функция, возвращающая тип объекта.
print("Тип:", type(b_data))   # <class 'bytes'>

# При индексации байтов получаем целое число (от 0 до 255).
print("Первый байт (число 72 — это 'H' в ASCII):", b_data[0])

text = "Привет"
# .encode("utf-8") превращает строку в байты по кодировке utf-8.
encoded = text.encode("utf-8")
print("Строка 'Привет' в байтах (utf-8):", encoded)

# .decode("utf-8") делает обратное: из байтов в строку.
decoded = encoded.decode("utf-8")
print("Обратно в строку:", decoded)

my_bytes = b"Python"   # поменяйте на b"12345"
print("Мои байты:", my_bytes)
print("Их длина:", len(my_bytes))
print("Третий байт (считаем с 0):", my_bytes[2])   # символ 't' (код 116)

# Байты можно складывать так же, как строки.
print("Конкатенация байтов:", b"Py" + b"thon")
print()

# Задание: закодировать своё имя:
# name_bytes = "Ваше имя".encode("utf-8")
# print(name_bytes)
```

🧊 11. frozenset — «Замороженное множество: как set, но неизменяемое»

frozenset — это точная копия set, но нельзя добавить, удалить или изменить элемент. Создаётся через frozenset().
Зачем? Чтобы можно было использовать множество как ключ в словаре или положить внутрь другого множества (обычный set нельзя).

```python
# Обычное множество: можно менять
normal_set = {1, 2, 3}
print("Обычное множество:", normal_set)
normal_set.add(4)          # добавляем элемент
print("После добавления 4:", normal_set)

# Создаём frozenset из списка (или из обычного множества)
frozen = frozenset([1, 2, 3, 3, 2])   # дубликаты автоматически убираются
print("Замороженное множество (frozenset):", frozen)
print("Тип:", type(frozen))

# Попытка изменить frozenset вызовет ошибку, раскомментируйте для проверки:
# frozen.add(4)   # AttributeError: 'frozenset' object has no attribute 'add'

# frozenset можно использовать как ключ в словаре (а обычный set — нельзя)
# Это как положить лёд в стакан: форма есть, но растопить нельзя.
dictionary = {
    frozenset([1, 2]): "пара чисел",
    frozenset([3]): "одиночка"
}
print("Словарь с ключами-frozenset:", dictionary)

# А вот попытка сделать ключом обычный set вызовет ошибку:
# dictionary_with_set = {{1, 2}: "ошибка"}   # TypeError: unhashable type: 'set'

# Создайте свой frozenset из строки "hello" (буквы превратятся в отдельные элементы)
my_frozen = frozenset("hello")
print("Мой frozenset из слова 'hello':", my_frozen)   # {'h', 'e', 'l', 'o'} порядок не важен

# Можно проверить, есть ли элемент внутри (как в обычном множестве)
print("Буква 'l' есть в my_frozen?", 'l' in my_frozen)   # True
print("Буква 'a' есть?", 'a' in my_frozen)              # False

# Длина frozenset (количество уникальных элементов)
print("Количество уникальных букв в 'hello':", len(my_frozen))
print()
```

🔄 12. bytearray — «Байты, которые можно менять (в отличие от bytes)»

bytearray — как bytes, но изменяемый. Можно менять отдельные байты, добавлять, удалять.
Представьте bytes как каменную табличку, а bytearray — как белую доску маркером.

```python
# Создаём bytearray из строки байтов (b'...') или из списка чисел от 0 до 255
ba = bytearray(b"Hello")
print("Исходный bytearray:", ba)           # bytearray(b'Hello')
print("Тип:", type(ba))

# Можно обращаться по индексу и менять значение (в отличие от bytes)
# Заменим 'H' (код 72) на 'J' (код 74)
ba[0] = 74                                 # 74 — это буква 'J' в ASCII
print("После замены первого символа:", ba)   # bytearray(b'Jello')

# Добавляем байт в конец (как в списке)
ba.append(33)                              # 33 — это восклицательный знак '!'
print("После добавления '!':", ba)          # bytearray(b'Jello!')

# Можно расширить другим bytearray или bytes
ba.extend(b" world")                       # добавляем последовательность байтов
print("После extend:", ba)                 # bytearray(b'Jello! world')

# Удаляем элемент по индексу (как в списке)
del ba[5]                                  # удаляем '!' (индекс 5)
print("Удалили восклицательный знак:", ba)  # bytearray(b'Jello world')

# Превратить bytearray обратно в строку (через декодирование)
decoded_str = ba.decode("utf-8")
print("В виде строки:", decoded_str)       # "Jello world"

# Создайте свой bytearray из чисел [65, 66, 67] (это A, B, C)
my_ba = bytearray([65, 66, 67])
print("Мой bytearray из чисел:", my_ba)     # bytearray(b'ABC')
# Поменяйте второй элемент (индекс 1) на 90 (буква Z)
my_ba[1] = 90
print("После замены B на Z:", my_ba)        # bytearray(b'AZC')
print()
```

🌀 13. range — «Ленивый генератор чисел (экономит память)»

range не хранит все числа в памяти, а генерирует их по требованию. Используется чаще всего в циклах for.
Выглядит как range(старт, стоп, шаг). старт по умолчанию 0, шаг по умолчанию 1.

```python
# Самый простой range: от 0 до 5 (не включая 5)
r = range(5)
print("range(5):", r)                # range(0, 5) — не числа, а обещание чисел
print("Тип:", type(r))

# Превратим range в список, чтобы увидеть все числа
print("Список из range(5):", list(r))   # [0, 1, 2, 3, 4]

# range с 2 аргументами: от 2 до 10 (не включая 10)
r2 = range(2, 10)
print("range(2, 10) в списке:", list(r2))   # [2, 3, 4, 5, 6, 7, 8, 9]

# range с 3 аргументами: старт, стоп, шаг (шаг может быть отрицательным)
r3 = range(1, 20, 3)
print("range(1, 20, 3) в списке:", list(r3))   # [1, 4, 7, 10, 13, 16, 19]

# Отрицательный шаг — идём назад
r4 = range(10, 0, -2)
print("range(10, 0, -2) в списке:", list(r4))  # [10, 8, 6, 4, 2]

# range не хранит все числа, поэтому даже огромный диапазон не съедает память
big_range = range(1_000_000_000)   # миллиард чисел
print("Размер большого range в памяти (байт):", big_range.__sizeof__())   # всего 48 байт!
# А если сделать список из миллиарда чисел — комп взорвётся (не делайте этого).

# Можно проверить, входит ли число в range (быстрая операция)
print("15 в range(10, 20)?", 15 in range(10, 20))   # True
print("25 в range(10, 20)?", 25 in range(10, 20))   # False

# Задание: создайте range, который выдаёт [5, 3, 1] (от 5 до 0 с шагом -2)
my_range = range(5, 0, -2)
print("Мой range:", list(my_range))   # [5, 3, 1]
print()
```

🧮 14. complex — «Комплексные числа (для математических гиков)»

complex — числа вида a + bj, где j — мнимая единица (квадрат = -1). Используются в науке, электротехнике, 3D-графике.
Обычные программисты могут не встретить их никогда, но знать полезно.

```python
# Комплексное число: действительная часть + мнимая часть с суффиксом j
z = 3 + 4j
print("Комплексное число:", z)
print("Тип:", type(z))
print("Действительная часть:", z.real)   # 3.0
print("Мнимая часть:", z.imag)           # 4.0
# Сопряжённое число (меняем знак мнимой части)
print("Сопряжённое:", z.conjugate())     # 3 - 4j

# Арифметика работает как в математике
z1 = 1 + 2j
z2 = 3 - 1j
print("z1 + z2 =", z1 + z2)   # (4+1j)
print("z1 * z2 =", z1 * z2)   # (5+5j) — умножение мнимых единиц даёт -1

# Можно создать комплексное число из двух чисел
z3 = complex(2, -3)   # 2 - 3j
print("complex(2, -3) =", z3)

# Модуль (расстояние от нуля на комплексной плоскости)
print("Модуль z =", abs(z))   # sqrt(3^2 + 4^2) = 5.0

# Задание: создайте число с действительной частью 0 и мнимой 1 (то есть 1j)
pure_imag = 1j
print("Квадрат мнимой единицы:", pure_imag ** 2)   # (-1+0j)
print()
```

```python
print("=" * 50)
print("1. ОСНОВНЫЕ ФУНКЦИИ ПРЕОБРАЗОВАНИЯ")
print("=" * 50)

# int() — превращает в целое число
print("int('42') =", int("42"))           # 42
print("int(3.99) =", int(3.99))           # 3 (дробная часть отбрасывается, не округляется)
print("int(True) =", int(True))           # 1 (True -> 1, False -> 0)

# float() — превращает в число с плавающей точкой
print("float('3.14') =", float("3.14"))   # 3.14
print("float(10) =", float(10))           # 10.0

# str() — превращает что угодно в строку
print("str(123) =", str(123))             # '123'
print("str(True) =", str(True))           # 'True'

# bool() — проверяет "истинность" значения
print("bool(0) =", bool(0))               # False (ноль — ложь)
print("bool(100) =", bool(100))           # True (любое ненулевое число — правда)
print("bool('') =", bool(""))             # False (пустая строка)
print("bool('Привет') =", bool("Привет")) # True (непустая строка)
print("bool([]) =", bool([]))             # False (пустой список)

print("\n" + "=" * 50)
print("2. ПРЕОБРАЗОВАНИЕ КОЛЛЕКЦИЙ")
print("=" * 50)

# list() — превращает итерируемый объект в список
text = "Metal"
print("list('Metal') =", list(text))      # ['M', 'e', 't', 'a', 'l']

# tuple() — превращает в кортеж (неизменяемый список)
print("tuple([1,2,3]) =", tuple([1,2,3])) # (1, 2, 3)

# set() — превращает в множество (убирает дубликаты)
numbers = [1, 2, 2, 3, 3, 3, 4]
print("set([1,2,2,3,3,3,4]) =", set(numbers)) # {1, 2, 3, 4}

# dict() — создаёт словарь из списка пар (кортежей)
pairs = [("client", "Иван"), ("income", 75000)]
print("dict(pairs) =", dict(pairs))       # {'client': 'Иван', 'income': 75000}

# Преобразование обратно: словарь -> список ключей/значений
d = {"name": "Ольга", "age": 29}
print("list(d.keys()) =", list(d.keys()))   # ['name', 'age']
print("list(d.values()) =", list(d.values())) # ['Ольга', 29]

print("\n" + "=" * 50)
print("3. ОБРАБОТКА ОШИБОК ПРИ КАСТИНГЕ")
print("=" * 50)

# Не все строки можно превратить в число
user_input = "зарплата"  # представим, что пользователь ввёл текст
try:
    value = int(user_input)
except ValueError as e:
    print(f"Ошибка! Не могу превратить '{user_input}' в число: {e}")

```

```python
# в данных иногда встречаются "ненормальные" строки
raw_income = "  85 000 руб."  # клиент ввёл доход с пробелами и текстом
# Убираем всё лишнее и кастим
cleaned = raw_income.replace("руб.", "").replace(" ", "").strip()
try:
    income = int(cleaned)
    print(f"Доход после очистки: {income} руб.")
except ValueError:
    print("Не удалось извлечь число")

print("\n" + "=" * 50)
print("4. PANDAS: astype() и pd.to_numeric()")
print("=" * 50)

import pandas as pd

# Создадим игрушечный датасет банковских клиентов
df = pd.DataFrame({
    'name': ['Иван', 'Ольга', 'Дмитрий'],
    'income': ['75000', '120000', 'пропуск'],  # колонка загрузилась как object (строка)
    'age': [30, 25, 40]
})

print("Исходные данные и типы:")
print(df)
print("\nТипы колонок до преобразования:\n", df.dtypes)

# Попытка прямого astype() выдаст ошибку из-за 'пропуск'
# Поэтому используем pd.to_numeric() с errors='coerce'
# coerce превращает всё "нечисловое" в NaN (Not a Number)
df['income_clean'] = pd.to_numeric(df['income'], errors='coerce')
print("\nПосле pd.to_numeric() с errors='coerce':")
print(df[['name', 'income', 'income_clean']])
print("Тип income_clean:", df['income_clean'].dtype)  # float64

# Теперь можно спокойно заменить NaN на медиану или удалить
median_income = df['income_clean'].median()
df['income_clean'] = df['income_clean'].fillna(median_income)
print("\nПосле заполнения пропусков медианой:")
print(df[['name', 'income_clean']])

# astype() для явного приведения: возраст из int64 -> float64
df['age_float'] = df['age'].astype('float64')
print("\nТип age_float:", df['age_float'].dtype)

print("\n" + "=" * 50)
print("5. БАНКОВСКИЙ КЕЙС: чистим и приводим типы")
print("=" * 50)

# Имитируем загрузку CSV с грязными данными
csv_data = """client,income,loan_amount
Сергей,95000,500000
Анна,неизвестно,300000
Алексей,120000,не заполнено
Мария,67000,250000
"""

from io import StringIO
df_raw = pd.read_csv(StringIO(csv_data))
print("Загруженный датасет и типы:")
print(df_raw)
print(df_raw.dtypes)

# Массовая очистка числовых колонок
for col in ['income', 'loan_amount']:
    df_raw[col] = pd.to_numeric(df_raw[col], errors='coerce')

print("\nПосле приведения типов (пропуски стали NaN):")
print(df_raw)
print(df_raw.dtypes)

# Заполним пропуски медианой (для учебного примера)
for col in ['income', 'loan_amount']:
    df_raw[col].fillna(df_raw[col].median(), inplace=True)

print("\nФинальный чистый датасет:")
print(df_raw)
```
