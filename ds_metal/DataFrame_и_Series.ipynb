## Создаём первый DataFrame из словаря

DataFrame — это просто таблица со строками и колонками. Представь, что ты ведёшь учёт клиентов, которые пришли за кредитом. У каждого клиента есть имя, доход, сумма кредита и решение: одобрить (True) или отказать (False).

В Python такую таблицу можно сделать из **словаря**. Ключи словаря станут названиями колонок, а значения (списки) — данными в строках.

```python
# Создаём обычный Python-словарь с данными клиентов
data = {
    'client': ["Cowboys from Hell", "Fear of the Dark", "True Belief"],
    # ↑ Колонка 'client' — имена клиентов (строки)
    'income': [75000, 120000, 55000],
    # ↑ Колонка 'income' — месячный доход в рублях (целые числа)
    'loan': [500000, 300000, 200000],
    # ↑ Колонка 'loan' — запрашиваемая сумма кредита
    'approved': [True, False, False]
    # ↑ Колонка 'approved' — одобрен ли кредит (логические True/False)
}

# Передаём словарь в pd.DataFrame()
# Pandas сам строит таблицу с автоматической нумерацией строк (0, 1, 2...)
df = pd.DataFrame(data)

# Выводим таблицу на экран
print(df)
```

## Series — одна колонка отдельно

Иногда нужна не вся таблица, а только одна колонка. Pandas хранит колонки как **Series** — одномерный набор данных с индексом.

Можно вытащить колонку из DataFrame или создать Series с нуля.

```python
# Способ 1: вытащить колонку из DataFrame по её названию
income_series = df['income']   # Берём колонку 'income'
print('Колонка income как Series:')
print(income_series)
print()  # пустая строка для читаемости

# Способ 2: создать Series вручную
scores = pd.Series(
    [0.89, 0.45, 0.22],          # список значений
    name='score',                 # название Series (необязательно)
    dtype='float64'               # тип данных: дробные числа
)
print('Series с нуля:')
print(scores)
```

## Смотрим на данные

Когда данных много, неудобно выводить всю таблицу целиком. В Pandas есть методы для быстрой "разведки":
- `.head(n)` — первые n строк (по умолчанию 5) , последние tail(n)
- `.info()` — информация о колонках, типах данных и памяти  
- `.shape` — кортеж (число строк, число колонок)

```python
# Создадим таблицу
clients = pd.DataFrame({
    'name': ['Ваня', 'Оля', 'Дима', 'Катя', 'Серёжа'],
    'income': [80000, 95000, 40000, 120000, 67000],
    'loan_amount': [300000, 2000000, 100000, 500000, 250000],
    'age': [29, 44, 19, 35, 51]
})

# .head() — первые строки (по умолчанию 5, но можно указать число)
print('Первые 3 строки таблицы clients:')
print(clients.head(3))
print()

print('последние 3 строки таблицы clients:')
print(clients.tail(3))
print()

# .info() — сводка о структуре таблицы
print('Информация о таблице clients:')
clients.info()
print()

# .shape — размер таблицы (строки, колонки)
print(f'Размер таблицы (строк, колонок): {clients.shape}')
```

## Метод `.describe()` — статистика за 5 сек

Хочешь быстро понять, какие доходы у твоих клиентов? Pandas даёт готовую статистику по числовым колонкам. Это как взглянуть на приборную панель мотоцикла: сразу видно скорость, температуру, пробег.

```python
# Метод .describe() считает основные статистики для каждой числовой колонки
print(clients.describe())
# Выведет: count (сколько непустых), mean (среднее), std (разброс), min, 25%, 50%, 75%, max

```

```python
# Если хочешь посмотреть только по конкретной колонке:
print('Средний доход:', clients['income'].mean())
print('Максимальный возраст:', clients['age'].max())
```

## Индексы — не просто номера

У каждой строки DataFrame есть индекс. По умолчанию это целые числа (0, 1, 2...), но ты можешь задать свой, например, id клиента. Индексы нужны для быстрого доступа к строкам и для объединения таблиц (будет в уроке 3.7).

"Twilight of the thunder god, Ragnarok awaits!" — Amon Amarth, Twilight of the Thunder God. Сейчас призовём гром среди строк.

```python
# Создадим DataFrame с осмысленным индексом (ID клиента)
clients_with_id = pd.DataFrame({
    'name': ['Ваня', 'Оля', 'Дима', 'Катя', 'Серёжа'],
    'income': [80000, 95000, 40000, 120000, 67000],
    'loan_amount': [300000, 2000000, 100000, 500000, 250000],
    'age': [29, 44, 19, 35, 51]
}, index=[101, 102, 103, 104, 105])   # <- задаём индексы явно

print(clients_with_id)
print()
print('Строка с индексом 103:')
print(clients_with_id.loc[103])   # доступ по значению индекса
```

## Типы данных колонок — .dtypes

Важно знать, как Pandas хранит твои данные. От этого зависит, какие операции можно делать.
- `int64` — целые числа (возраст, доход)  
- `float64` — дробные числа (процентная ставка)  
- `bool` — правда/ложь (одобрен кредит?)  
- `string` — текст (имя клиента) — в Pandas 3.0 это Arrow-based

```python
print('Типы данных в таблице clients_with_id:')
print(clients_with_id.dtypes)
```

## Создание DataFrame из списка словарей (как JSON)

Часто данные приходят из API банка в виде списка словарей — например, ответ на запрос "дай мне клиентов". Pandas умеет превращать такой список в таблицу мгновенно.

```python
# Список словарей — как будто нам прислали JSON с сервера
json_like_data = [
    {'name': 'Ваня', 'income': 80000, 'loan': 300000, 'approved': True},
    {'name': 'Оля', 'income': 95000, 'loan': 2000000, 'approved': False},
    {'name': 'Дима', 'income': 40000, 'loan': 100000, 'approved': False}
]

# Просто передаём список в pd.DataFrame()
df_from_json = pd.DataFrame(json_like_data)
print(df_from_json)
# Pandas сам поймёт, что ключи — это колонки, а значения — строки
```

## Добавляем новую колонку и вычисляем показатель долговой нагрузки (DTI)

В банке часто считают **DTI (debt-to-income)** — доля кредита от годового дохода. DTI = сумма кредита / (доход * 12). Если DTI < 0.3 — хорошо, >0.5 — рискованно.

Добавим такую колонку в DataFrame. Pandas позволяет делать это одной строкой, как мощный рифф.

```python
# Берём clients_with_id и создаём новую колонку 'dti'
clients_with_id['dti'] = clients_with_id['loan_amount'] / (clients_with_id['income'] * 12)

print('Таблица с DTI:')
print(clients_with_id)

# А теперь посмотрим только на рискованных клиентов (DTI > 0.5)
risky = clients_with_id[clients_with_id['dti'] > 0.5]
print('\nКлиенты с высоким риском (DTI > 0.5):')
print(risky)
```

## Copy-on-Write на практике: цепочки операций

Начиная с Pandas 3.0, при фильтрации или изменении части данных **оригинал защищён**. Если берёшь кусок таблицы и меняешь его, Pandas автоматически создаёт копию, чтобы не испортить исходные данные. Это называется **Copy-on-Write**.

```python
# Исходная таблица
df_original = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'balance': [1000, 2000, 1500, 3000]
})
print('Оригинал:')
print(df_original)

# Выберем подмножество и попробуем изменить баланс
subset = df_original[df_original['balance'] > 1000]  # все, у кого баланс > 1000
subset['balance'] = 0  # в Pandas 3.0 это создаст копию

print('\nПосле изменения подмножества:')
print('Подмножество:')
print(subset)
print('Оригинал (не изменился!):')
print(df_original)
# Вывод: изменения в subset НЕ затронули df_original
```
