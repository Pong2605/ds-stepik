🌀 for — перебиратель всего подряд

Когда использовать: когда знаете, ПО ЧЕМУ бежать (по списку, строке, диапазону чисел).

```python
# Перебираем список песен
playlist = ["Bohemian Rhapsody", "Stairway to Heaven", "Smells Like Teen Spirit"]
for song in playlist:
    # song по очереди принимает каждое значение из списка
    print(f"🎸 Играем: {song}")

# Перебираем строку по буквам
riff = "RIFF"
for char in riff:
    print(char)  # R, I, F, F

# Используем range(старт, стоп, шаг) для чисел
for i in range(3):        # 0,1,2
    print(f"Такт {i}")
for i in range(1, 6, 2):  # 1,3,5
    print(i)
```

🔄 while — крутись, пока условие верно

Когда использовать: когда не знаете, сколько раз крутиться, но знаете, когда остановиться.

```python
# Секундомер
seconds = 3
while seconds > 0:
    print(f"Осталось {seconds} секунд")
    seconds -= 1   # уменьшаем на 1, иначе бесконечный цикл!
print("Рванули!")

# Бесконечный цикл с выходом по break
while True:
    answer = input("Ты рокер? (да/нет): ")
    if answer == "да":
        print("Добро пожаловать!")
        break
    else:
        print("Попробуй ещё раз...")
```

🛑 break — стоп-кран

```python
# Ищем первую песню короче 4 букв
songs = ["Hey Jude", "U2", "Led Zeppelin", "Yes"]
for s in songs:
    if len(s) <= 3:
        print(f"Нашёл короткую: {s}")
        break   # вылетаем из цикла, не проверяя остальные
# Вывод: U2 (дальше не идём)
```

⏭️ continue — перемотать этот шаг

```python
for i in range(5):
    if i == 2:
        continue   # пропускаем двойку, но не выходим из цикла
    print(i)   # 0,1,3,4
```

🎁 else у цикла — бонусная дорожка

Блок else выполняется, если цикл закончился не по break, а «естественно».

```python
# Поиск злой собаки
dogs = ["Бобик", "Шарик", "Тузик"]
for dog in dogs:
    if dog == "Рекс":
        print("Осторожно, Рекс кусается!")
        break
else:
    print("Все собаки добрые, можно гулять.")  # напечатается, потому что Рекса нет
```

🗒️ List comprehension — списки на лету

```python
# Обычный цикл
squares = []
for x in range(5):
    squares.append(x**2)

# Генератор списка (то же самое, но красиво)
squares = [x**2 for x in range(5)]   # [0,1,4,9,16]

# С условием
even_squares = [x**2 for x in range(10) if x % 2 == 0]
# [0,4,16,36,64]
```

📖 Dict comprehension — словари

```python
# Словарь: число -> его куб
cubes = {x: x**3 for x in range(1, 4)}   # {1:1, 2:8, 3:27}
```

🎲 Set comprehension — множества

```python
unique_lengths = {len(word) for word in ["a", "bb", "ccc", "bb"]}
# {1,2,3}
```

# map — применить функцию к каждому элементу

```python
nums = [1,2,3,4]
doubled = list(map(lambda x: x*2, nums))   # [2,4,6,8]

# Или с обычной функцией
def double(x): return x*2
doubled = list(map(double, nums))
```

# filter — оставить только те, которые подходят

```python
nums = [1,2,3,4,5]
evens = list(filter(lambda x: x % 2 == 0, nums))   # [2,4]
```

# reduce — свернуть список в одно значение

```python
from functools import reduce
nums = [1,2,3,4]
total = reduce(lambda a, b: a + b, nums)   # 10 (1+2+3+4)
# Можно и произведение: lambda a,b: a*b
```

# enumerate — бегу с индексом

Когда в цикле нужно знать не только значение, но и его порядковый номер.

```python
songs = ["Песня года", "Хит лета", "Баллада"]
for idx, song in enumerate(songs):
    print(f"{idx+1}. {song}")
# 1. Песня года
# 2. Хит лета
# 3. Баллада

# Можно начать с другого числа: enumerate(songs, start=1)
```

# Рекурсия — когда функция вызывает саму себя

```python
# Факториал через рекурсию
def factorial(n):
    if n <= 1:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(5))   # 120

# Обратный отсчёт
def countdown(n):
    if n <= 0:
        print("Бум!")
    else:
        print(n)
        countdown(n-1)
countdown(3)
```

# Итераторы и генераторы (yield)

Итератор — это объект, по которому можно пройтись for. Генератор — это функция с yield, которая возвращает значения по одному, экономя память.

```python
# Генератор бесконечной последовательности
def rock_numbers():
    n = 0
    while True:
        yield n   # возвращаем значение, но функция не завершается
        n += 1

# Использование
gen = rock_numbers()
print(next(gen))   # 0
print(next(gen))   # 1
# Но можно и в for:
for r in rock_numbers():
    if r > 5: break
    print(r, end=" ")   # 0 1 2 3 4 5
```

Модуль itertools — швейцарский нож циклов

```python
from itertools import cycle, repeat, product, permutations  # Подключаем из встроенной библиотеки только нужные функции, чтобы писать их имена напрямую, без приставки "itertools."

colors = cycle(["красный", "жёлтый", "зелёный"])  # Создаём "бесконечный генератор": он будет по кругу отдавать цвета из списка, никогда не останавливаясь
for i in range(7):  # Запускаем обычный цикл, который выполнится ровно 7 раз. Переменная i здесь выступает просто счётчиком, её значение нам не нужно
    print(next(colors))  # Функция next() "вытаскивает" следующий цвет из генератора и сразу передаёт его в print() для вывода на экран

for _ in repeat("🤘", 3):  # Функция repeat() создаёт последовательность, которая повторит смайлик 3 раза. Знак _ — это соглашение Python: "значение мне не нужно, важно лишь количество повторов"
    print("Рок жив!")  # На каждом из 3 шагов цикла выводим эту строку в консоль

for a, b in product([1,2], ["A","B"]):  # Функция product() комбинирует каждый элемент первого списка со всеми элементами второго (заменяет два вложенных цикла одной строкой)
    print(a, b)  # Выводим получившуюся пару, например: 1 A, затем 1 B, затем 2 A и т.д.

for p in permutations([1,2,3], 2):  # Функция permutations() генерирует все возможные упорядоченные комбинации. Цифра 2 означает, что в каждой комбинации будет по два элемента из исходного списка
    print(p)  # Выводим каждую комбинацию на экран (она возвращается в виде кортежа, например (1, 2))
```


next() нужен, потому что cycle отдаёт значения по одному и "запоминает", где остановился. Без next() вы бы получили просто объект-генератор, а не сам цвет.


_ вместо имени переменной (for _ in ...) — это стандартный трюк, который означает "мне не важно, что возвращает цикл, я использую его только для повторения действий".

# Вложенные циклы — «песня в песне»

Внутри цикла может быть ещё один цикл. Например, печатаем таблицу умножения:

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} * {j} = {i*j}")
    print("---")
```
