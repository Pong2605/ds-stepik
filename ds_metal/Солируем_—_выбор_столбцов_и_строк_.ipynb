```python
import pandas as pd
import numpy as np

np.random.seed(42)
N = 1000
names = ['Иван', 'Ольга', 'Дмитрий', 'Екатерина', 'Сергей', 'Анна', 'Алексей', 'Мария', 'Николай', 'Татьяна']

df = pd.DataFrame({
    'client_id': range(1001, 1001 + N),
    'name': np.random.choice(names, N),
    'age': np.random.randint(21, 66, N),
    'income': np.random.randint(20000, 500001, N),
    'dti': np.round(np.random.uniform(0, 1, N), 2),
    'default': np.random.choice([True, False], N, p=[0.15, 0.85]),
    'application_date': pd.date_range('2024-01-01', periods=N, freq='h')
}).set_index('client_id')  # делаем client_id индексом для удобства loc

print(df.shape)
print(df.head())
```

## loc — выбор по меткам

loc позволяет обращаться к строкам по значению индекса и к колонкам по названиям.

```python
# Одна строка по индексу (например, client_id=1005)
print("Строка с индексом 1005:")
print(df.loc[1005])

# Несколько строк по списку индексов
print("\nСтроки с индексами 1005 и 1010:")
print(df.loc[[1005, 1010]])

# Диапазон индексов (включая конец!)
print("\nСтроки с 1005 по 1009 (включая 1009):")
print(df.loc[1005:1009])

# Конкретные строки и колонки
print("\nСтроки 1005-1007, колонки name и income:")
print(df.loc[1005:1007, ['name', 'income']])

# Все строки, но только колонки от 'age' до 'dti' (срез по именам колонок)
print("\nВсе строки, колонки от 'age' до 'dti':")
print(df.loc[:, 'age':'dti'].head())
```

## iloc — выбор по позициям (номерам)

iloc работает с целочисленными позициями, как индексы в списке.

```python
# Первая строка (позиция 0)
print("Первая строка:")
print(df.iloc[0])

# Строки с 5 по 9 (не включая 10!)
print("\nСтроки 5-9:")
print(df.iloc[5:10])

# Строки 5-9 и колонки 0-2 (колонки нумеруются с 0)
print("\nСтроки 5-9, колонки 0-2:")
print(df.iloc[5:10, 0:3])

# Последняя колонка для всех строк
print("\nПоследняя колонка:")
print(df.iloc[:, -1].head())

# Каждая сотая строка (шаг)
print("\nКаждая сотая строка, колонки 1 и 3:")
print(df.iloc[::100, [1, 3]])
```

##  Обычный доступ через квадратные скобки

Самый простой способ взять одну или несколько колонок.

```python
# Одна колонка -> Series
income_series = df['income']
print("Тип дохода:", type(income_series))
print(income_series.head())

# Несколько колонок -> DataFrame
subset = df[['name', 'income', 'dti']]
print("\nТип subset:", type(subset))
print(subset.head())

# Выбор строк через срез (не рекомендуется, лучше iloc)
print("\nСтроки 0-4 (как iloc):")
print(df[0:5])

# Булева фильтрация (превью следующего урока)
print("\nТолько клиенты с доходом > 300000:")
print(df[df['income'] > 300000].head())
```

## pd.col()

pd.col() позволяет ссылаться на колонки в цепочках методов.

```python
# pd.col НЕ требует df.select — он работает напрямую в .assign() и .loc[]

# 1. Создание новой колонки через .assign()
df_with_flag = df.assign(
    high_income = pd.col('income') > 200000
)
print("Пример assign с pd.col:")
print(df_with_flag[['income', 'high_income']].head())

# 2. Фильтрация через .loc[]
high_dti = df.loc[pd.col('dti') > 0.5]
print(f"\nКлиентов с DTI > 0.5: {len(high_dti)}")

# 3. Группировка (pd.col не работает в agg, используем кортежи)
result = df.groupby('default').agg(
    avg_income=('income', 'mean'),
    max_age=('age', 'max'),
    total_clients=('income', 'count')
)
print("\nГруппировка (совместимый синтаксис):")
print(result)
```

## Сравнение loc и iloc

loc включает конечную метку в срезе, iloc — НЕТ. Проверим на практике.

```python
# loc: срез [1005:1009] включает 1009
print("loc[1005:1009]:")
print(df.loc[1005:1009])

# iloc: срез [5:9] НЕ включает 9 (т.е. строки 5,6,7,8)
print("\niloc[5:9]:")
print(df.iloc[5:9])

# если у тебя индекс — числа, не перепутай
print("\nПример: df.loc[0] ищет СТРОКУ С ИНДЕКСОМ 0")
print("df.iloc[0] ищет ПЕРВУЮ строку")
```

## Комбинированная выборка

Извлеки только клиентов с высоким доходом (> 300 000), оставив только имя и DTI.

```python
# Фильтруем и выбираем колонки
high_income_clients = df.loc[df['income'] > 300000, ['name', 'dti']]
print(f"Клиентов с доходом > 300 000: {len(high_income_clients)}")
print(high_income_clients.head(10))
```

## Выбор по условию в индексе

Найди всех клиентов с id от 1200 до 1210 (индекс — client_id).

```python
# Индекс у нас — client_id
clients_1200_1210 = df.loc[1200:1210]  # loc включает правую границу
print(clients_1200_1210)
```

## Срез с шагом

Возьми каждую 200-ю строку (по позиции) и колонки 'name', 'income'.

```python
# Используем iloc с шагом
every_200th = df.iloc[::200, [0, 2]]  # колонки: name (0) и income (2)
print(every_200th)
```

## «Кредитный конвейер»

Начальник попросил тебя сделать выборку для пилотного кредитного продукта:
- только клиенты от 30 до 50 лет (включительно)
- доход от 100 000 до 300 000
- столбцы: name, age, income, dti
- отсортировать по доходу от большего к меньшему

```python
pilot = df.loc[
    (df['age'] >= 30) & (df['age'] <= 50) &
    (df['income'] >= 100000) & (df['income'] <= 300000),
    ['name', 'age', 'income', 'dti']
].sort_values('income', ascending=False)

print(f"Отобрано клиентов: {len(pilot)}")
print(pilot.head(10))
```
