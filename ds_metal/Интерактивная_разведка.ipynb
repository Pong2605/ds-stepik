```python
# Импорт библиотек
import pandas as pd
import numpy as np
import dtale

print("Библиотеки загружены!")
```

```python
# Генерация банковского датасета (как в предыдущих уроках)
np.random.seed(42)
N = 2000
cities = ['Москва', 'Санкт-Петербург', 'Казань', 'Новосибирск', 'Екатеринбург']
df = pd.DataFrame({
    'client_id': range(1001, 1001 + N),
    'city': np.random.choice(cities, N),
    'age': np.random.randint(21, 70, N),
    'income': np.random.normal(50000, 20000, N).astype(int),
    'loan_amount': np.random.normal(200000, 80000, N).astype(int),
    'credit_score': np.random.normal(650, 100, N).astype(int),
    'default': np.random.choice([0, 1], N, p=[0.9, 0.1])
})
df['income'] = df['income'].clip(lower=0)
df['loan_amount'] = df['loan_amount'].clip(lower=0)
print(f"Датасет: {df.shape[0]} строк, {df.shape[1]} колонок")
df.head()
```

## D‑Tale — панель управления датасетом

D‑Tale запускает локальный веб‑сервер прямо в Colab и предоставляет графический интерфейс для Pandas DataFrame.

**Как это работает:**
- `dtale.show(df)` — запускает сервер и отображает таблицу с меню.
- В открывшемся окне можно сортировать, фильтровать, строить графики и получать автоматический код на Python.

```python
import dtale.app as dtale_app
dtale_app.USE_COLAB = True
```

```python
dtale.show(df)
```

**Навигация в D‑Tale:**
- **Column menu** (стрелка рядом с названием колонки): описать, отсортировать, фильтровать.
- **Верхнее меню**: Actions → Visualize → выбирай график.
- **Correlations**: тепловая карта корреляций с подсветкой.
- **Missing**: анализ пропусков.
- **Code Export**: кнопка "</>" показывает код для воспроизведения графика.

**Банковский пример:** через меню Visualize построй scatter plot `income` vs `loan_amount`, раскрась по `default`, и ты сразу увидишь группы дефолтников без написания кода.

## 🧩 PyGWalker — Tableau внутри Colab

PyGWalker превращает DataFrame в drag‑and‑drop исследование данных. Ты можешь перетаскивать поля на полки (столбцы, строки, цвет, размер) и мгновенно видеть результат.

```python
import pygwalker as pyg
walker = pyg.walk(df)
```

**Инструкция к PyGWalker:**
1. Слева — список полей (колонок).
2. Перетащи поле в **Rows** (строки) или **Columns** (колонки) — появится график.
3. Меняй тип графика в верхней панели (bar, scatter, line, histogram, heatmap и др.).
4. Добавляй размер, цвет через перетаскивание.
5. Кнопка "Export" сохраняет график как PNG или HTML.

**Банковский пример:**
- Перетащи `income` на Columns, `loan_amount` на Rows.
- В Color перетащи `default` — сразу увидишь, где дефолтники.
- Можешь добавить `city` в Size, чтобы увидеть города с крупными кредитами.
