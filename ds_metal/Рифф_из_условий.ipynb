📀 Раздел 1: if-elif-else — главный дирижёр

Простейший if

```python
# Переменная — количество зрителей на концерте
audience = 1000

# if — "если". Проверяем: зрителей больше 500?
if audience > 500:
    # Если да, печатаем "Ого, аншлаг!"
    print("Ого, аншлаг!")
# Обратите внимание: после if ставим двоеточие и отступ.
# Если условие ложно (audience <= 500), print не выполнится — программа просто пройдёт мимо.
print("Программа продолжается...")  # эта строка вне if, выполнится всегда
```

Поиграйте: поменяйте audience на 300 и запустите снова. Что изменилось?

if-else — развилка

```python
temperature = 28  # градусов на улице

if temperature > 25:
    # Если жарко
    action = "Пить лимонад"
else:
    # Иначе — прохладно
    action = "Пить чай"

print("Сегодня:", action)

# Расшифровка: else — "иначе". Выполнится, когда условие if ложно.
# Попробуйте temperature = 20 — увидите "Пить чай".
```

if-elif-else — много вариантов

```python
score = 85  # баллы за гитарное соло

if score >= 90:
    level = "Легенда"
elif score >= 75:
    level = "Мастер"
elif score >= 60:
    level = "Ученик"
else:
    level = "Начало пути"

print(f"Ваш уровень: {level}")

# Как работает: Python идёт сверху вниз, как только находит истинное условие — выполняет его и выходит.
# score=85 попадает во второй вариант (85>=75), остальные не проверяются.
# Измените score на 55 — попадёте в "Ученик". На 95 — "Легенда".
```

Логические операторы and, or, not

```python
age = 20
has_ticket = True

# Условие: возраст от 18 до 30 И (and) есть билет
# and требует, чтобы оба условия были True
if age >= 18 and age <= 30 and has_ticket:
    print("Проходите в VIP-зону!")
else:
    print("Доступ закрыт.")

# Поэкспериментируйте:
# - поменяйте age=17 или has_ticket=False
# - используйте or (замените and на or) — тогда достаточно одного условия

# not — "не"
is_weekend = False
if not is_weekend:
    print("Сегодня рабочий день — но мы всё равно репетируем!")
```

Вложенные условия (условие внутри условия)

```python
has_money = True
has_guitar = True

if has_money:
    # Если деньги есть, проверяем, есть ли гитара
    if has_guitar:
        print("Покупаем новую гитару! 🎸")
    else:
        print("Копим на гитару.")
else:
    print("Идём играть на улице бесплатно.")

# Можно писать условия любой глубины, но не увлекайтесь — становится сложно читать.
```

Базовый тернарник

```python
# Синтаксис: значение_если_True if условие else значение_если_False
tempo = 130
genre = "Металл" if tempo > 120 else "Рок"
print(genre)  # → Металл

# Это то же самое, что:
if tempo > 120:
    genre = "Металл"
else:
    genre = "Рок"
# Но короче. Используйте для простых случаев.
```

Тернарник с присваиванием в функцию

```python
def get_fee(hours):
    # Если часов больше 2, плата 100, иначе 50
    return 100 if hours > 2 else 50

print(get_fee(3))  # 100
print(get_fee(1))  # 50
```

Простой match-case как замена if-elif

```python
command = "play"

match command:
    case "play":
        print("🎸 Воспроизведение трека")
    case "pause":
        print("⏸️ Пауза")
    case "stop":
        print("⏹️ Остановлено")
    case _:   # аналог else — всё остальное
        print("Неизвестная команда")

# Попробуйте поменять command на "pause", "stop", "rewind"
```

match-case с несколькими вариантами и условиями

```python
people = 4

match people:
    case 1:
        desc = "Солист"
    case 2:
        desc = "Дуэт"
    case 3 | 4:      # можно перечислять через |
        desc = "Группа"
    case n if n > 4: # с условием (guard)
        desc = f"Оркестр из {n} человек"
    case _:
        desc = "Пустая сцена"

print(desc)  # Группа
```
