```python
import matplotlib.pyplot as plt
import numpy as np

# Логическое И

# Пример коэффициентов, удовлетворяющих таблице истинности функции **И**
w_1 = 2
w_2 = 1
b = -2

# Наша (разделяющая) линия: w_1 * x_1 + w_2 * x_2 + b = 0
def line(x1, x2):
    return w_1 * x1 + w_2 * x2 + b

# Служебная функция в форме x_2 = f(x_1) (для наглядности):
def line_x1(x1):
    return (-w_1 * x1 - b) / w_2

# Генерируем наши точки
x1x2 = np.array([[0,0],[1,0],[0,1],[1,1]])

# Рисуем точки
for x1, x2 in x1x2:
    value = line(x1, x2)
    
    if (value == 0): # синие — на линии
        plt.plot(x1, x2, 'o', color='blue')
    elif (value > 0): # зелёные — выше линии
        plt.plot(x1, x2, '^', color='green')
    elif (value < 0): # красные — ниже линии
        plt.plot(x1, x2, '^', color='red')

# Рисуем линию
x1_range = np.arange(-0.1, 1.5, 0.1)
plt.plot(x1_range, line_x1(x1_range), color='blue')

# Проставляем названия осей, а также сетку
plt.xlabel('x1')
plt.ylabel('x2')
plt.axis('equal')
plt.grid(which='minor', alpha=0.1)
plt.grid(which='major', alpha=0.2)

# Выводим график на экран
plt.show()
```

```python
# Логическое ИЛИ

# Пример коэффициентов, удовлетворяющих таблице истинности функции **ИЛИ**
w_1 = 1.5
w_2 = 0.75
b = -1

# Наша (разделяющая) линия: w_1 * x_1 + w_2 * x_2 + b = 0


def line(x1, x2):
    return w_1 * x1 + w_2 * x2 + b

# Служебная функция в форме x_2 = f(x_1) (для наглядности):


def line_x1(x1):
    return (-w_1 * x1 - b) / w_2


# Генерируем наши точки
x1x2 = np.array([[0, 0], [1, 0], [0, 1], [1, 1]])

# Рисуем точки
for x1, x2 in x1x2:
    value = line(x1, x2)
    
    if (value >= 0):  # зелёные — выше или на линии
        plt.plot(x1, x2, '^', color='green')
    elif (value < 0):  # красные — ниже линии
        plt.plot(x1, x2, '^', color='red')

# Рисуем линию
x1_range = np.arange(-0.1, 1.5, 0.1)
plt.plot(x1_range, line_x1(x1_range), color='blue')

# Проставляем названия осей, а также сетку
plt.xlabel('x1')
plt.ylabel('x2')
plt.axis('equal')
plt.grid(which='minor', alpha=0.1)
plt.grid(which='major', alpha=0.2)

# Выводим график на экран
plt.show()
```

```python
# Логическое XOR

# Пример коэффициентов, удовлетворяющих таблице истинности функции **XOR**
w1_1 = 1
w2_1 = -1
b_1 = -0.5

w1_2 = -1
w2_2 = 1
b_2 = -0.5

# Наша (разделяющая) линия: w_1 * x_1 + w_2 * x_2 + b = 0


def line_1(x1, x2):
    return w1_1 * x1 + w2_1 * x2 + b_1


def line_2(x1, x2):
    return w1_2 * x1 + w2_2 * x2 + b_2

# Служебная функция в форме x_2 = f(x_1) (для наглядности):

def line_x1_1(x1):
    return (-w1_1 * x1 - b_1) / w2_1


def line_x1_2(x1):
    return (-w1_2 * x1 - b_2) / w2_2


# Генерируем наши точки
x1x2 = np.array([[0, 0], [1, 0], [0, 1], [1, 1]])

# Рисуем точки
for x1, x2 in x1x2:
    value_1 = line_1(x1, x2)
    value_2 = line_2(x1, x2)
    
    if not (value_1 >= 0) and (value_2 <= 0):  # зелёные — выше или на линии
        plt.plot(x1, x2, '^', color='green')
    else:  # красные — ниже линии
        plt.plot(x1, x2, '^', color='red')

# Рисуем линии
x1_range = np.arange(-0.5, 2.0, 0.1)
plt.plot(x1_range, line_x1_1(x1_range), color='blue', label='value_1')

plt.plot(x1_range, line_x1_2(x1_range), color='orange', label='value_2')

# Проставляем названия осей, а также сетку
plt.xlabel('x1')
plt.ylabel('x2')
plt.axis('equal')
plt.legend()
plt.grid(which='minor', alpha=0.1)
plt.grid(which='major', alpha=0.2)

# Выводим график на экран
plt.show()
```

```python
def xor(a, b):
    return int((a and not b) or (not a and b))


def neuron(x):
    return 1 if x > 0 else 0


input = [(0, 0), (0, 1), (1, 0), (1, 1)]

print('Реализация XOR через xor:')
for x1, x2 in input:
    print(x1, x2, xor(x1, x2))

print('\nРеализация НЕ через neuron:')
for x in [0, 1]:
    print(x, neuron(-x + 0.5))

print('\nРеализация И через neuron:')   # 1 1 -1.5
for x1, x2 in input:
    print(x1, x2, neuron(x1 + x2 - 1.5))

print('\nРеализация ИЛИ через neuron:')  # 1 1 -0.5
for x1, x2 in input:
    print(x1, x2, neuron(x1 + x2 - 0.5))

print('\nРеализация XOR через neuron:')
for x1, x2 in input:
    print(x1, x2, neuron(neuron(x1 + x2 - 0.5) -
          neuron(x1 + x2 - 1.5) - 0.5))   # (ИЛИ) - (И) - 0.5
```
